Exploration only: every function used here lives in `src/conso_forecast/`, so the

```python
import pandas as pd

from conso_forecast.config import Settings
from conso_forecast.data.clean import load_clean
from conso_forecast.eda import calendar_effects, daily_table, summarize

settings = Settings()
df = load_clean(settings)
df.describe()
```

## Missing values and data quality

```python
quality = pd.read_json(settings.paths.results / "data_quality.json", typ="series")
quality
```

```python
df.isna().groupby(df.index.year).sum()
```

## Seasonality

```python
daily = daily_table(df)
daily["consumption_mw"].plot(figsize=(12, 4), linewidth=0.6, title="Daily mean consumption (MW)");
```

```python
df.assign(month=df.index.month, hour=df.index.hour).pivot_table(
    index="hour", columns="month", values="consumption_mw", aggfunc="mean"
).plot(figsize=(10, 5), title="Mean profile by hour, one line per month");
```

## Calendar effects

```python
calendar_effects(daily)
```

## Temperature effect

```python
daily.plot.scatter(
    x="temperature_c", y="consumption_mw", c="month", cmap="twilight", s=6, figsize=(7, 5)
);
```

## Summary used in the report

```python
summarize(df)
```
